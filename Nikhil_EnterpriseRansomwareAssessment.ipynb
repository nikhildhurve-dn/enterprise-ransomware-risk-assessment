## 1. Objectives
- Identify plausible ransomware risk scenarios.
- Score likelihood and impact on a 1–5 scale.
- Prioritize risks using likelihood × impact.
- Map recommendations to NIST CSF 2.0 functions.
- Export a risk register for review.

```python
import pandas as pd
from pathlib import Path

OUTPUT_DIR = Path(".")
```

## 2. Risk scoring method
Likelihood and impact are scored from 1 (low) to 5 (very high). The risk score is their product. Thresholds used here are project-defined: 1–5 Low, 6–14 Medium, 15–25 High; they are not universal NIST ratings.

```python
risk_data = [
    {"Risk ID":"R01","Risk Scenario":"Phishing email enables credential theft or malware execution","Assets":"Email users and endpoints","Likelihood":4,"Impact":5,"Controls":"Email filtering; awareness training; MFA; reporting process","NIST CSF Mapping":"Protect / Detect","Owner":"IT Security Lead","Target":"30 days"},
    {"Risk ID":"R02","Risk Scenario":"Unpatched software is exploited","Assets":"Servers and workstations","Likelihood":4,"Impact":4,"Controls":"Asset inventory; risk-based patching; vulnerability management","NIST CSF Mapping":"Identify / Protect","Owner":"IT Operations","Target":"30 days"},
    {"Risk ID":"R03","Risk Scenario":"Ransomware spreads laterally across the network","Assets":"Endpoints, file shares, servers","Likelihood":3,"Impact":5,"Controls":"Network segmentation; endpoint detection; least privilege","NIST CSF Mapping":"Protect / Detect","Owner":"Network Administrator","Target":"60 days"},
    {"Risk ID":"R04","Risk Scenario":"Backups are encrypted, deleted, or unavailable","Assets":"Backup repositories and business data","Likelihood":3,"Impact":5,"Controls":"Offline/immutable backups; separate credentials; restore tests","NIST CSF Mapping":"Protect / Recover","Owner":"Backup Administrator","Target":"30 days"},
    {"Risk ID":"R05","Risk Scenario":"Excessive privileges increase compromise impact","Assets":"Identity and access management","Likelihood":3,"Impact":4,"Controls":"Least privilege; separate admin accounts; access reviews; MFA","NIST CSF Mapping":"Govern / Protect","Owner":"IAM Administrator","Target":"45 days"},
    {"Risk ID":"R06","Risk Scenario":"Suspicious activity is not detected promptly","Assets":"Logs, endpoints, identity systems","Likelihood":3,"Impact":4,"Controls":"Centralized logging; alerting; incident triage","NIST CSF Mapping":"Detect","Owner":"Security Analyst","Target":"45 days"},
    {"Risk ID":"R07","Risk Scenario":"Incident response roles and communications are unclear","Assets":"Business operations and response team","Likelihood":3,"Impact":4,"Controls":"Written response plan; contacts; tabletop exercise","NIST CSF Mapping":"Govern / Respond","Owner":"Incident Response Coordinator","Target":"30 days"},
    {"Risk ID":"R08","Risk Scenario":"Sensitive data is exposed during or after an incident","Assets":"Customer and business data","Likelihood":2,"Impact":5,"Controls":"Data classification; encryption; access controls; escalation path","NIST CSF Mapping":"Identify / Protect / Respond","Owner":"Data Protection Owner","Target":"60 days"},
    {"Risk ID":"R09","Risk Scenario":"Third-party remote access is abused","Assets":"Vendor access and remote administration","Likelihood":3,"Impact":4,"Controls":"MFA; time-limited access; vendor reviews; approved gateway","NIST CSF Mapping":"Govern / Protect","Owner":"Vendor Risk Owner","Target":"60 days"},
    {"Risk ID":"R10","Risk Scenario":"Recovery is delayed because restoration priorities are unknown","Assets":"Critical business services","Likelihood":3,"Impact":4,"Controls":"Business impact analysis; recovery priorities; tested recovery targets","NIST CSF Mapping":"Govern / Recover","Owner":"Business Continuity Lead","Target":"60 days"},
]
df = pd.DataFrame(risk_data)
df["Risk Score"] = df["Likelihood"] * df["Impact"]

def risk_level(score):
    if score >= 15:
        return "High"
    if score >= 6:
        return "Medium"
    return "Low"

df["Risk Level"] = df["Risk Score"].apply(risk_level)
df["Residual Risk (illustrative)"] = ["Medium","Medium","Medium","Low-Medium","Medium","Medium","Medium","Medium","Medium","Medium"]
df
```

## 3. Review the highest-priority risks
The following cell sorts the risk register from highest to lowest score. These are illustrative scores for a hypothetical company, not measured results from a real audit.

```python
priority_view = df.sort_values(["Risk Score", "Risk ID"], ascending=[False, True])
priority_view[["Risk ID","Risk Scenario","Likelihood","Impact","Risk Score","Risk Level","Owner","Target"]]
```

## 4. NIST CSF 2.0 function mapping
- **Govern:** policies, responsibilities, third-party risk, oversight.
- **Identify:** assets, data, dependencies, business impact.
- **Protect:** MFA, least privilege, patching, endpoint protection, backups.
- **Detect:** logs, monitoring, alert triage.
- **Respond:** incident roles, containment, communications.
- **Recover:** validated backups, restoration priorities, recovery testing.

This is a high-level mapping, not a formal control-by-control compliance assessment.

```python
csf_mapping = pd.DataFrame([
    {"NIST CSF 2.0 Function":"Govern","Example actions":"Set policy, assign owners, review third-party access and risk."},
    {"NIST CSF 2.0 Function":"Identify","Example actions":"Maintain asset inventory and identify critical business services."},
    {"NIST CSF 2.0 Function":"Protect","Example actions":"Use MFA, least privilege, patching, segmentation and resilient backups."},
    {"NIST CSF 2.0 Function":"Detect","Example actions":"Centralize logs and alert on suspicious activity."},
    {"NIST CSF 2.0 Function":"Respond","Example actions":"Coordinate incident analysis, containment and communications."},
    {"NIST CSF 2.0 Function":"Recover","Example actions":"Restore trusted systems from verified backups and review lessons learned."},
])
csf_mapping
```
